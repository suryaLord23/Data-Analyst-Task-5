# 🚢 Titanic Dataset - Exploratory Data Analysis (EDA)

**Objective**: Extract insights using visual and statistical exploration.

**Tools**: Python (Pandas, Matplotlib, Seaborn)

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("titanic dataset.csv")
df.head()
```

```
df.info()
```

```
df.describe(include='all')
```

## 🧹 Data Cleaning

```
df_cleaned = df.drop(columns=['Unnamed: 12', 'Cabin', 'Ticket', 'Cleaned_Name'])
df_cleaned['Age'].fillna(df_cleaned['Age'].median(), inplace=True)
df_cleaned['Fare'].fillna(df_cleaned['Fare'].median(), inplace=True)
```

## 📊 Univariate Analysis

```
sns.set(style="whitegrid")
fig, axes = plt.subplots(3, 2, figsize=(14, 15))
fig.suptitle('Univariate Analysis', fontsize=18)

sns.histplot(df_cleaned['Age'], bins=30, kde=True, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title("Age Distribution")

sns.histplot(df_cleaned['Fare'], bins=30, kde=True, ax=axes[0, 1], color='salmon')
axes[0, 1].set_title("Fare Distribution")

sns.countplot(x='Survived', data=df_cleaned, ax=axes[1, 0], palette='pastel')
axes[1, 0].set_title("Survival Count")

sns.countplot(x='Sex', data=df_cleaned, ax=axes[1, 1], palette='muted')
axes[1, 1].set_title("Gender Distribution")

sns.countplot(x='Pclass', data=df_cleaned, ax=axes[2, 0], palette='deep')
axes[2, 0].set_title("Passenger Class Distribution")

sns.countplot(x='Embarked', data=df_cleaned, ax=axes[2, 1], palette='Set2')
axes[2, 1].set_title("Port of Embarkation")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

## 🔍 Bivariate Analysis

```
fig, axes = plt.subplots(3, 2, figsize=(14, 15))
fig.suptitle('Bivariate Analysis: Relationship with Survival', fontsize=18)

sns.countplot(x='Sex', hue='Survived', data=df_cleaned, ax=axes[0, 0], palette='pastel')
axes[0, 0].set_title("Survival by Gender")

sns.countplot(x='Pclass', hue='Survived', data=df_cleaned, ax=axes[0, 1], palette='muted')
axes[0, 1].set_title("Survival by Passenger Class")

sns.countplot(x='Embarked', hue='Survived', data=df_cleaned, ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title("Survival by Embarkation Port")

sns.boxplot(x='Survived', y='Age', data=df_cleaned, ax=axes[1, 1], palette='coolwarm')
axes[1, 1].set_title("Age Distribution by Survival")

sns.boxplot(x='Survived', y='Fare', data=df_cleaned, ax=axes[2, 0], palette='coolwarm')
axes[2, 0].set_title("Fare Distribution by Survival")

corr = df_cleaned.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='Blues', ax=axes[2, 1])
axes[2, 1].set_title("Correlation Heatmap")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

## 📝 Summary of Findings

- **Females** and **1st class passengers** had a much higher survival rate.
- Passengers from **Cherbourg** had better outcomes than other ports.
- **Younger individuals** had higher chances of survival.
- Higher **fare** correlated with better survival outcomes.
- **Pclass**, **Fare**, and **Sex** were the most influential factors.
